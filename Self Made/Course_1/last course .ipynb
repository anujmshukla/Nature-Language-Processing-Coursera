```python
def make_basic_hashtable(value_list, n_bucket):

    def get_hash_value(value, n_bucket):
        return value % n_bucket

    hash_table = {i:[] for i in range(n_bucket)}
    for val in value_list:
        hash_value = get_hash_value(val, n_bucket)
        hash_table[hash_value].append(val)

    return hash_table
```

```python
P = np.array([[1, 1]])      # Single plane
v1 = np.array([[1, 2]])     # Sample point 1
v2 = np.array([[-1, 1]])    # Sample point 2
v3 = np.array([[-2, -1]])   # Sample point 3
```

```python
def get_side_of_plane(p,v):
    dot = np.dot(p,v.T)
    sign = np.sign(dot)
    return sign
```

```python
get_side_of_plane(P, v1) # In which side is [1, 2]
```

```python
P1 = np.array([[1, 1]])   # First plane 2D
P2 = np.array([[-1, 1]])  # Second plane 2D
P3 = np.array([[-1, -1]]) # Third plane 2D
P_l = [P1, P2, P3]  # List of arrays. It is the multi plane

# Vector to search
v = np.array([[2, 2]])
```

```python
def get_combine_hash_value(p_l,v):
    hash_value = 0
    for i,p in enumerate(p_l):
        side = get_side_of_plane(p,v)
        ind_hash = 1 if side >-1 else 0
        hash_value += (2**i)*ind_hash
    return(hash_value)
```

```python
get_combine_hash_value(P_l, v)
```

```python
def hash_multi_plane_matrix(P, v, num_planes):
    sides_matrix = get_side_of_plane(P, v) # Get the side of planes for P and v
    hash_value = 0
    for i in range(num_planes):
        sign = sides_matrix[i].item() # Get the value inside the matrix cell
        hash_i = 1 if sign >=0 else 0
        hash_value += 2**i * hash_i # sum 2^i * hash_i
        
    return hash_value
```

```python
hash_multi_plane_matrix(random_plane, v, 3)
```
