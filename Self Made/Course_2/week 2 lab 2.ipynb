```python
tag = ['RB','NN','TO']
```

```python
transition_counts = {
    ('NN', 'NN'): 16241,
    ('RB', 'RB'): 2263,
    ('TO', 'TO'): 2,
    ('NN', 'TO'): 5256,
    ('RB', 'TO'): 855,
    ('TO', 'NN'): 734,
    ('NN', 'RB'): 2431,
    ('RB', 'NN'): 358,
    ('TO', 'RB'): 200
}
```

```python
num_tag = len(tag)
trans_matrix = np.zeros((num_tag,num_tag))
trans_matrix
```

```python
sorted_tag = sorted(tag)
sorted_tag
```

```python
for i in range(len(sorted_tag)):
    for j in range(len(sorted_tag)):
        get_tag = (sorted_tag[i],sorted_tag[j])
        trans_matrix[i, j] = transition_counts.get(get_tag)

trans_matrix
```

```python
trans_matrix =  pd.DataFrame(trans_matrix,index=sorted_tag, columns=sorted_tag)
trans_matrix
```
