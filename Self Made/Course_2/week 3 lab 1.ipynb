```python
import pandas as pd 
import numpy as np
```

```python
def get_count_matrix(Cropus):

    bigrams = [] 
    vocabs = []
    count_matrix_dict = {}

    for i in range(len(Cropus)-3+1):

        trigram = tuple(Cropus[i:i+3])
        # print(trigram)

        bigram = trigram[0:-1]
        if not bigram in bigrams:
            bigrams.append(bigram)
        
        vocab = trigram[2]
        if not vocab in vocabs:
            vocabs.append(vocab)

        if (bigram,vocab) in count_matrix_dict:
            count_matrix_dict[(bigram,vocab)] += 1
        else:
            count_matrix_dict[(bigram,vocab)] = 1

    # print(count_matrix_dict)
    count_matrix = np.zeros((len(bigrams),len(vocabs)))
    for keys, count in count_matrix_dict.items():
        # print(keys,count)
        # print(bigrams.index(keys[0]))
        count_matrix[bigrams.index(keys[0]),vocabs.index(keys[1]) ] = count

    count_matrix = pd.DataFrame(count_matrix, index=bigrams, columns=vocabs)
    return count_matrix


corpus = ['i', 'am', 'happy', 'because', 'i', 'am', 'learning', '.']
count_matrix = get_count_matrix(corpus)
count_matrix
```

```python
row_sum = count_matrix.sum(axis=1)
row_sum
```

```python
Prob_matrix = count_matrix.div(row_sum, axis=0)
```
