```python
# Now let's make the function to conver the line into tensor
def line_to_tensor(line,vocab):
    chars = tf.strings.unicode_split(line,input_encoding='UTF-8')
    embedding = tf.keras.layers.StringLookup(vocabulary=list(vocab),mask_token=None)(chars)
    return embedding
```

```python
#testing the function
tmp_ids = line_to_tensor('abc xyz', vocab)
print(f"Result: {tmp_ids}")
print(f"Output type: {type(tmp_ids)}")
# tf.keras.layers.StringLookup(vocabulary=list(vocab), invert=True, mask_token= None)(tmp_ids)
```

```python
# making reverse function i.e from embedding to text 
def tensor_to_line(tensor,vocab):
    char_from_ids= tf.keras.layers.StringLookup(vocabulary=list(vocab), invert=True, mask_token= None)
    line = tf.strings.reduce_join(char_from_ids(tensor))
    return line
```

```python
def split_input_target(seq):
    input = seq[:-1]
    output = seq[1:]
    return input, output
```

```python
split_input_target(list("Tensorflow"))
```

```python
def create_batch_dataset(lines, vocab, seq_length=100, batch_size=64):
    shuffler = 10000
    combile_line = "\n".join(lines)
    tensor_line = line_to_tensor(combile_line,vocab)
    id_dataset = tf.data.Dataset.from_tensor_slices(tensor_line)
    data_generator = id_dataset.batch(sequence_length+1,drop_remainder=True)
    data_inp_out = data_generator.map(split_input_target)

    dataset = (
        data_inp_out
        .shuffle(10000)
        .batch(batch_size, drop_remainder=True)
        .prefetch(tf.data.experimental.AUTOTUNE) 
    )
    return dataset
```

```python
# test your function
dataset = create_batch_dataset(train_line[1:100], vocab, seq_length=16, batch_size=2)

print("Prints the elements into a single batch. The batch contains 2 elements: ")

for input_example, target_example in dataset.take(1):
    print("\n\033[94mInput0\t:", tensor_to_line(input_example[0], vocab).numpy())
    print("\n\033[93mTarget0\t:", tensor_to_line(target_example[0], vocab).numpy())
    
    print("\n\n\033[94mInput1\t:", tensor_to_line(input_example[1], vocab).numpy())
    print("\n\033[93mTarget1\t:", tensor_to_line(target_example[1], vocab).numpy())
```

```python
#creating dataset
batch_size = 64
dataset = create_batch_dataset(train_line,vocab,batch_size=batch_size,seq_length=100)
```

```python
vocab_len = len(vocab)
embedding = 256
rnn_unit = 512
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(vocab_len,embedding),
    tf.keras.layers.GRU(rnn_unit,return_sequences=True),
    tf.keras.layers.Dense(vocab_len,activation=tf.nn.log_softmax)
])
```

```python
model.build(input_shape=(batch_size,100))
model.summary()
```

```python
for input, output in dataset.take(1):
    print(tensor_to_line(input[0],vocab))
    print(input[0])
    pridection = model(tf.constant([input[0].numpy()]))
    print(pridection.shape,"(batch_size, sequence_length, vocab_size)")
    print(vocab[np.argmax(pridection[0][-1].numpy())])
# model is prodecting anything 
```

```python
# we have to train the model 
model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer= tf.keras.optimizers.Adam(learning_rate=0.00125)
)
```

```python
history = model.fit(
    dataset,
    epochs=10,
    verbose= True,
)
```
